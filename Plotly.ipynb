```python
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import plotly.io as pio
```

```python
months = ['January', 'February', 'March', 'April', 'May','June']
sales = [100, 1500, 2500, 500, 2000, 2000]
df= pd.DataFrame({"months": months,"sales": sales})
```

```python
plt.plot(df['months'], df['sales'], color='blue', marker='o')
plt.xlabel('months')
plt.ylabel('sales')
plt.title('Monthly Sales')
plt.grid()
plt.legend()
plt.show()
```

```python
sns.lineplot(x=months, y=sales, data=df, color='blue', marker='o', label='sales')
plt.xlabel('months')
plt.ylabel('sales')
plt.title('Monthly Sales')
plt.grid()
plt.legend()
plt.show()
```

```python
sns.barplot(x=months, y=sales, palette='Blues_d')
plt.xtricks
plt.xlabel('months')
plt.ylabel('sales')
plt.title('Monthly Sales')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.hist(sales, bins= 10)

plt.xlabel('months')
plt.ylabel('sales')
plt.title('Monthly Sales')
plt.grid()
plt.legend()
plt.show()
```
