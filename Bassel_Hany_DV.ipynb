```python
import matplotlib.pyplot as plt
import numpy as np
```

<span style="font-size:45px;">1. Line Plot</span>


```python
months = np.arange(1, 7)
sales_2025 = [120, 150, 170, 160, 210, 240]
sales_2026 = [140, 165, 190, 200, 230, 275]
```

```python
plt.plot(months, sales_2025, color='blue', marker='o', label='2025 Sales')
plt.plot(months,sales_2026,color='red',linestyle='--',linewidth=2,marker='s',label='2026 Sales',)

plt.title('Monthly Sales Comparison')
plt.xlabel('Month')
plt.ylabel('Revenue ($1000s)')

plt.grid(True)
plt.legend(loc='upper left')
```

<span style="font-size:45px;">2. Bar Chart</span>


```python
courses = ['Python', 'Data Science', 'AI', 'Web Dev', 'Cybersecurity']
students = [85, 120, 95, 60, 75]

plt.figure(figsize=(8, 5))

plt.bar(
    courses, 
    students, 
    width=0.6, 
    color='teal', 
    edgecolor='black', 
    label='2026 Intake'
)

plt.title('Student Enrollment per Track')
plt.xlabel('Course Track')
plt.ylabel('Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

<span style="font-size:45px;">3. Area Plot</span>


```python
days = np.arange(1, 8)
active_users = [200, 240, 310, 280, 390, 450, 500]

plt.figure(figsize=(8, 5))

plt.fill_between(
    days, 
    active_users, 
    color='skyblue', 
    alpha=0.5, 
    label='Daily Active Users'
)
plt.plot(
    days, 
    active_users, 
    color='navy', 
    linewidth=2
)

plt.title('Weekly App Traffic (Area Chart)')
plt.xlabel('Day of the Week')
plt.ylabel('Users Count')
plt.grid(True, alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

<span style="font-size:45px;">4. Histogram</span>


```python
np.random.seed(42)
scores = np.random.normal(loc=75, scale=10, size=500)

plt.figure(figsize=(8, 5))

# 2. Plot histogram
plt.hist(
    scores, 
    bins=20, 
    color='purple', 
    edgecolor='white', 
    alpha=0.8, 
    label='Score Distribution'
)

plt.axvline(
    np.mean(scores), 
    color='gold', 
    linestyle='--', 
    linewidth=2, 
    label='Mean Score'
)

plt.title('Distribution of Final Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.4)
plt.legend()

plt.tight_layout()
plt.show()
```

<span style="font-size:45px;">5. Scatter Plot</span>


```python
study_hrs_A = [1, 2, 3, 4, 5, 6, 7]
grades_A = [50, 55, 65, 70, 78, 85, 90]
study_hrs_B = [1, 2, 3, 4, 5, 6, 7]
grades_B = [45, 50, 58, 62, 68, 72, 79]

plt.figure(figsize=(8, 5))
plt.scatter(
    study_hrs_A, 
    grades_A, 
    color='green', 
    marker='o', 
    s=80, 
    label='Morning Group'
)
plt.scatter(
    study_hrs_B, 
    grades_B, 
    color='darkorange', 
    marker='^', 
    s=80, 
    label='Evening Group'
)
plt.title('Study Hours vs. Exam Performance')
plt.xlabel('Study Hours / Day')
plt.ylabel('Grade (%)')
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

<span style="font-size:45px;">6. Pie Chart</span>


```python
languages = ['Python', 'JavaScript', 'C++', 'Java', 'Other']
usage = [45, 25, 12, 10, 8]
colors = ['#2b5b84', '#f7df1e', '#00599c', '#f89820', '#94a3b8']
explode = (0.1, 0, 0, 0, 0)  # Explode the first slice ('Python')

plt.figure(figsize=(8, 6))
plt.pie(
    usage, 
    labels=languages, 
    colors=colors, 
    explode=explode, 
    autopct='%1.1f%%', 
    shadow=True, 
    startangle=140
)
plt.title('Programming Language Market Share')
plt.legend(loc='best')
plt.axis('equal') 
plt.tight_layout()
plt.show()
```

<span style="font-size:45px;">7. Subplots</span>


```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
fig, axes = plt.subplots(2, 1, figsize=(8, 6))

axes[0].plot(x, y1, color='crimson', label='sin(x)')
axes[0].set_title('Sine Wave')
axes[0].set_ylabel('Amplitude')
axes[0].grid(True, alpha=0.4)
axes[0].legend()

axes[1].plot(x, y2, color='darkgreen', linestyle='--', label='cos(x)')
axes[1].set_title('Cosine Wave')
axes[1].set_xlabel('Time (s)')
axes[1].set_ylabel('Amplitude')
axes[1].grid(True, alpha=0.4)
axes[1].legend()

plt.tight_layout()
plt.show()
```
