```python
import matplotlib.pyplot as plt
```

```python
months=[1,2,3,4,5]
sales=[10,15,12,5,20]
```

```python
months2=[6,7,8,9,10]
sales2=[12,9,10,6,15]
```

<span style="font-size:45px;">1. Line Chart</span>


```python
plt.plot(months, sales)
plt.show()
```

```python
plt.plot(months, sales, color='red', marker='o')
plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.show()
```

```python
plt.plot(months, sales, color='red', marker='o')
plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.grid()
plt.title("Sales in the first five months")
plt.show()
```

```python
plt.plot(months, sales, color='red', marker='o', label='Sales')
plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.grid(linestyle='--')
plt.title("Sales in the first five months")
plt.legend()
plt.show()
```

```python
plt.plot(months, sales, color='red', marker='o', label='Sales', linestyle='--')
plt.plot(months2, sales2, color='green', marker='*', label='Sales for the rest')
plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.grid(linestyle='--')
plt.title("Sales in the first five months")
plt.legend()
plt.show()
```

<span style="font-size:45px;">2. Area Chart</span>


```python
plt.fill_between(months, sales, color='red', label='Sales', linestyle='--')
plt.fill_between(months2, sales2, color='green', label='Sales for the rest')

plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.grid(linestyle='--')
plt.title("Sales in the first five months")
plt.legend()
plt.show()
```

<span style="font-size:45px;">3. Bar Chart</span>


```python
players=["CR7", "Polga", "Benzima"]
pallondor=[5,8,1]
```

```python
plt.bar(players, pallondor, color=['r', 'g', 'y'], label= ["CR7","Polga","Benzima"], linestyle='--')

plt.xlabel("The Player")
plt.ylabel("Pallon")
plt.grid(linestyle='--')
plt.title("Graph on Pallon")
plt.legend()
plt.show()
```

```python
plt.bar(months, sales, color='red', label='Sales', linestyle='--')
plt.bar(months2, sales2, color='green', label='Sales for the rest')

plt.xlabel("The first five months")
plt.ylabel("Sales in $")
plt.grid(linestyle='--')
plt.title("Sales in the first five months")
plt.legend()
plt.show()
```

<span style="font-size:45px;">4. Scatter Plot</span>


```python
plt.scatter(players, pallondor,marker='*', color=['r', 'g', 'y'], label= ["CR7","Polga","Benzima"])

plt.xlabel("The Player")
plt.ylabel("Pallon")
plt.grid(linestyle='--')
plt.title("Graph on Pallon")
plt.legend()
plt.show()
```

<span style="font-size:45px;">5. Bubble Plot</span>


```python
players=["CR7", "Polga", "Benzima"]
pallondor=[5,8,1]
buble= [1000,500,1500]
```

```python
plt.scatter(players, pallondor, s=buble, color=['r', 'g', 'y'], label= ["CR7","Polga","Benzima"])

plt.xlabel("The Player")
plt.ylabel("Pallon")
plt.grid(linestyle='--')
plt.title("Graph on Pallon")
plt.legend()
plt.show()
```

<span style="font-size:45px;">6. Histogram</span>


```python
students=[20, 30, 25, 20, 15, 10, 20, 35, 30, 10, 5]
```

```python
plt.hist(students, bins=7, color='purple',edgecolor='white')
plt.ylabel('Number of students')
plt.xlabel('Ages')
plt.title('Distribution of ages')
plt.show()
```

<span style="font-size:45px;">7. Pie Chart</span>


```python
plt.pie(students,autopct='%1.1f%%', shadow=True)
```
